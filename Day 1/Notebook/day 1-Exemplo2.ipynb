```python
import requests
import json
from datetime import datetime
```

```python
base_url = 'https://api.exchangeratesapi.io/v1'
endpoint = '/latest'
access_key = '25f2eddc12596495ba4916982abbfc15'
data = datetime.now().strftime("%Y-%m-%d")

url = base_url + endpoint + '?access_key=' + access_key
print("Endpoint:           " + url)
print("Data Extração       " + data)
```

```python
#Realizar GET request
response = requests.get(url)

#verificar a requisição
if response.status_code == 200:
    json_data = response.json()
    print(f"Requisição bem-sucedida!")
    print(f"Dados recebidos: {list(json_data.keys())}\n")
    #display(json_data)
else:
    print(f"Requisição falhou com código de status: {response.status_code}")
    print(response.text)
```
