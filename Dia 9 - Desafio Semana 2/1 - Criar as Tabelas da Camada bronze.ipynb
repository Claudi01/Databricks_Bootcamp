```python
%sql
CREATE OR REPLACE TABLE `desafio-semana-2`.bronze.bronze_clientes
COMMENT 'Tabela bronze com os dados brutos de clientes' AS
SELECT
    try_cast(id_cliente AS STRING) AS id_cliente,
    try_cast(nome AS STRING) AS nome,
    try_cast(sexo AS STRING) AS sexo,
    try_cast(data_nascimento AS STRING) AS data_nascimento,
    try_cast(cidade AS STRING) AS cidade,
    try_cast(estado AS STRING) AS estado,
    try_cast(data_cadastro AS STRING) AS data_cadastro,
    from_utc_timestamp(current_timestamp(), 'America/Sao_Paulo') AS data_ingestao,
    'clientes.csv' AS arquivo_origem
FROM read_files('/Volumes/desafio-semana-2/landing/files/clientes.csv',
    format => 'csv', header => true);
```

```python
%sql
CREATE OR REPLACE TABLE `desafio-semana-2`.bronze.bronze_suporte
COMMENT 'Tabela bronze com os dados brutos de suporte' AS
SELECT
    try_cast(id_interacao AS STRING) AS id_interacao,
    try_cast(id_cliente AS STRING) AS id_cliente,
    try_cast(canal AS STRING) AS canal,
    try_cast(tipo_problema AS STRING) AS tipo_problema,
    try_cast(tempo_resolucao AS STRING) AS tempo_resolucao,
    try_cast(satisfacao_cliente AS STRING) AS satisfacao_cliente,
    from_utc_timestamp(current_timestamp(), 'America/Sao_Paulo') AS data_ingestao,
    'suporte.csv' AS arquivo_origem
FROM read_files('/Volumes/desafio-semana-2/landing/files/suporte.csv',
    format => 'csv', header => true);
```

```python
%sql
CREATE OR REPLACE TABLE `desafio-semana-2`.bronze.bronze_vendas
COMMENT 'Tabela bronze com os dados brutos de vendas' AS
SELECT
    try_cast(id_venda AS STRING) AS id_venda,
    try_cast(id_cliente AS STRING) AS id_cliente,
    try_cast(data_venda AS STRING) AS data_venda,
    try_cast(produto AS STRING) AS produto,
    try_cast(quantidade AS STRING) AS quantidade,
    try_cast(valor_total AS STRING) AS valor_total,
    from_utc_timestamp(current_timestamp(), 'America/Sao_Paulo') AS data_ingestao,
    'vendas.csv' AS arquivo_origem
FROM read_files('/Volumes/desafio-semana-2/landing/files/vendas.csv',
    format => 'csv', header => true);
```

```python
%sql
SELECT 'clientes' AS tabela, COUNT(*) AS linhas FROM `desafio-semana-2`.bronze.bronze_clientes
UNION ALL SELECT 'suporte', COUNT(*) FROM `desafio-semana-2`.bronze.bronze_suporte
UNION ALL SELECT 'vendas',   COUNT(*) FROM `desafio-semana-2`.bronze.bronze_vendas;
```
