```python
%sql
CREATE TABLE IF NOT EXISTS `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.silver.sales
USING DELTA
AS 
SELECT 
  s.`Order_ID` AS id_venda, 
  CAST(s.`Order_Time` AS STRING) AS data_venda_formatada, 
  c.`Customer_ID` AS id_cliente,
  UPPER(c.`Customer_Name`) AS nome_cliente, 
  p.`Product_ID` AS id_produto,
  COALESCE(p.`Category`, 'Sem Categoria') AS categoria_produto, 
  CAST(s.`Quantity` AS INT) AS quantidade,
  CAST(s.`Unit_Price` AS DECIMAL(10,2)) AS preco_unitario
FROM `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.bronze.sales s
LEFT JOIN `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.bronze.customers c 
  ON s.`Customer_ID` = c.`Customer_ID`
LEFT JOIN `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.bronze.products p 
  ON s.`Product_ID` = p.`Product_ID`
WHERE 1=0;
```

```python
%sql
MERGE INTO `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.silver.sales AS destino
USING (
  SELECT 
    s.`Order_ID` AS id_venda, 
    CAST(s.`Order_Time` AS STRING) AS data_venda_formatada,             
    c.`Customer_ID` AS id_cliente,
    UPPER(c.`Customer_Name`) AS nome_cliente,                          
    p.`Product_ID` AS id_produto,
    COALESCE(p.`Category`, 'Sem Categoria') AS categoria_produto,     
    CAST(s.`Quantity` AS INT) AS quantidade,
    CAST(s.`Unit_Price` AS DECIMAL(10,2)) AS preco_unitario
  FROM `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.bronze.sales s
  LEFT JOIN `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.bronze.customers c 
    ON s.`Customer_ID` = c.`Customer_ID`
  LEFT JOIN `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.bronze.products p 
    ON s.`Product_ID` = p.`Product_ID`
  WHERE CAST(s.`Quantity` AS INT) > 0                                 
) AS origem
ON destino.id_venda = origem.id_venda
WHEN MATCHED THEN
  UPDATE SET *
WHEN NOT MATCHED THEN
  INSERT *;
```
