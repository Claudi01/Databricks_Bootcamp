```python
%sql
CREATE TABLE IF NOT EXISTS `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.gold.indicador_vendas
USING DELTA
AS
SELECT
  categoria_produto,
  SUM(quantidade) AS total_produtos_vendidos,
  SUM(quantidade * preco_unitario) AS faturamento_total
FROM `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.silver.sales
GROUP BY categoria_produto
ORDER BY faturamento_total DESC;
```

```python
%sql
CREATE TABLE IF NOT EXISTS `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.gold.ranking_clientes
USING DELTA
AS
SELECT
  nome_cliente,
  SUM(quantidade * preco_unitario) AS valor_gasto_total,
  COUNT(DISTINCT id_venda) AS total_pedidos
FROM `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.silver.sales
GROUP BY nome_cliente
ORDER BY valor_gasto_total DESC
LIMIT 10;
```

```python
%sql
CREATE TABLE IF NOT EXISTS `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.gold.bonus_horarios_pico
USING DELTA
AS
SELECT
  SUBSTRING(data_venda_formatada, 1, 2) AS hora_do_dia,
  SUM(quantidade) AS total_itens_vendidos,
  SUM(quantidade * preco_unitario) AS faturamento_hora
FROM `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.silver.sales
GROUP BY SUBSTRING(data_venda_formatada, 1, 2)
ORDER BY faturamento_hora DESC;
```

```python
%sql
CREATE OR REPLACE VIEW `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.gold.vw_top_clientes 
AS
SELECT 
  nome_cliente, 
  valor_gasto_total
FROM `capgemini-academy-challenge-LuizClaudioPinheirodeMouraFilho`.gold.ranking_clientes;
```
