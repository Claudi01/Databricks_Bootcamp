```sql
CREATE TABLE IF NOT EXISTS capgemini_academy.silver.movie_ratings_summary AS
    SELECT
      m.movieId,
      m.title,
      m.genres,
      COUNT(r.rating) AS total_avaliacoes,
      AVG(r.rating) AS media_avaliacoes
    FROM
      capgemini_academy.bronze.movies AS m
      JOIN capgemini_academy.bronze.ratings AS r
        ON (m.movieId = r.movieId)
    GROUP BY
      m.movieId,
      m.title,
      m.genres;
```
