```sql
------------------------------------------------
CREATE OR REPLACE TABLE capgemini_academy.bronze.links(
  movieId INT COMMENT 'Movie ID from MovieLens',
  imdbId INT COMMENT 'IMDB ID',
  tmdbId INT COMMENT 'The Movie Database (TMDb) ID',
  updated TIMESTAMP COMMENT 'Timestamp of last update'
)
COMMENT 'Tabela bronze com links dos filmes do MovieLens para IMDb e TMDb';

------------------------------------------------
SELECT movieId, imdbId, tmdbId,
from_utc_timestamp(current_timestamp(), 'America/Sao_Paulo') AS updated
FROM read_files(
  '/Volumes/capgemini_academy/landing/files/movielens/links.csv',
  format => 'csv',
  header => true
);
```
