```sql
TRUNCATE TABLE `capgemini_academy`.`bronze`.`exchange_rates`;
------------------------------------
INSERT INTO `capgemini_academy`.`bronze`.`exchange_rates`
    SELECT
        base,
        data_base,
        moeda,
        valor
    FROM (
        SELECT
            from_utc_timestamp(from_unixtime(timestamp), 'America/Sao_Paulo') AS data_extracao,
            base,
            TO_DATE(date) AS data_base,
            rates
        FROM read_files(
            '/Volumes/capgemini_academy/landing/files/exchangerates/*.json',
            format => 'json',
            multiLine => true)
    ) 
    LATERAL VIEW EXPLODE(FROM_JSON(TO_JSON(rates), 'MAP<STRING,DOUBLE>')) AS moeda, valor;
------------------------------------
```
