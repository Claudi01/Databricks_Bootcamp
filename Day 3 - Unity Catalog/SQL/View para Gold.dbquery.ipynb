```sql
---------------------------------------------
CREATE OR REPLACE VIEW `capgemini_academy`.`gold`.`vw_top_rated_movies` AS
    SELECT
        movieId,
        title,
        genres,
        total_avaliacoes,
        media_avaliacoes
    FROM `capgemini_academy`.`silver`.`movie_ratings_summary`
    WHERE total_avaliacoes >= 50
    ORDER BY media_avaliacoes DESC;
---------------------------------------------
```
