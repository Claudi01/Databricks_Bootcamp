```python
%sql
SELECT
    from_utc_timestamp(from_unixtime(timestamp), 'America/Sao_Paulo') AS data_estracao,
    base,
    TO_DATE(date) AS data_base,
    rates,
    _metadata.file_path AS file
FROM read_files(
    '/Volumes/capgemini_academy/landing/files/exchangerates/exchange_rates_2026-09-16.json',
    format => 'json',
    multiline => true);
```

```python
%sql
CREATE OR REPLACE TABLE capgemini_academy.bronze.exchange_rates AS
SELECT 
  base, 
  date, 
  explode(from_json(to_json(rates), 'MAP<STRING, DOUBLE>')) AS (currency, exchange_rate)
FROM read_files(
  '/Volumes/capgemini_academy/landing/files/exchangerates/*.json',
  format => 'json',
  multiline => true
);

SELECT * FROM capgemini_academy.bronze.exchange_rates;
```
